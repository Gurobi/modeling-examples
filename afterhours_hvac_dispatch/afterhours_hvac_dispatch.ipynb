# After-Hours HVAC Dispatch

Assign on-call HVAC technicians to after-hours emergency jobs with time windows.
A missed night call that hits voicemail is a lost job. This MIP maximizes captured job value on the size-limited pip `gurobipy` license.

Motivating application: after-hours missed-call recovery ([AfterHours Ops](https://github.com/IgorGanapolsky/ai-operations-agency)). Teaching model only.

```python
%pip install -q gurobipy
import gurobipy as gp
from gurobipy import GRB
from afterhours_hvac_dispatch import build_and_solve, TECHS, JOBS

m, x = build_and_solve()
assert m.Status == GRB.OPTIMAL
print('captured_job_value', m.ObjVal)
for t in TECHS:
    print(t, [j['id'] for j in JOBS if x[t, j['id']].X > 0.5] or 'idle')
```
